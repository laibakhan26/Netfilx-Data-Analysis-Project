```python
#Lets just make a clear dataframe to find how many movie id are there
#for every row in CustID we will run a loop and select the row where we have colon
#and then store it in y after replacing colon with nothing "".
movie_np=[]
for x in df['CustID']:
  if ':' in x:
    y=x.replace(':',"")
  movie_np.append(y)
```

```python
movie_np
```

```python
df['MovieID']=movie_np
```

```python
df
```

```python
# To keep only the rows where the rating column is not null ( example It exclude rows whete the Rating is NaN)
df.dropna(inplace=True)
```

```python
df
```

```python
df.info()
```

```python
df['CustID']=df['CustID'].astype(int)
df['MovieID']=df['MovieID'].astype(int)
```

```python
df.info()
```

```python
#Manual Filteration
#I want to get the count of every customer with respect to ratings
data_cust_summary=df.groupby('CustID')['Ratings'].count()
data_cust_summary
```

```python
cust_benchmark=round(data_cust_summary.quantile(0.60))
# I will select only those customers who have seen more than 36 movies.
cust_benchmark
```

```python
rejected_cust=data_cust_summary[data_cust_summary<cust_benchmark].index
# When we use groupby fxn the column becomes index
rejected_cust
```

```python
#Filtering our movie data
#First we will get the count of ratings for evry movie.
data_movie_summary=df.groupby('MovieID')['Ratings'].count()
```

```python
data_movie_summary
```

```python
movie_benchmark=round(data_movie_summary.quantile(0.60))
movie_benchmark
```

```python
rejected_movies=data_movie_summary[data_movie_summary<movie_benchmark].index
rejected_movies
```

```python
# Now we have selected the customers and movies that we need to reject.
# Now we will update the result in our actual dataframe (removing these rejected values)
```

```python
len(rejected_cust)
```

```python
#So this line keeps only the rows where the movie ID cust_ID is NOT in the rejected_movies and rejected_cust list.
df=df[~df['MovieID'].isin(rejected_movies)]
df=df[~df['CustID'].isin(rejected_cust)]
```
